## Case Study 1 – AI Pathogen Surveillance Model  
*(ONT, Underrepresented Pathogens & Low-Quality Reads)*

A research team wants to build an AI model that predicts which pathogen is present in citrus metagenomic samples.  
For training, they use:

- **Low-quality ONT reads**, many with Q < 10  
- **Uneven, unbalanced data**  
  - Some pathogens have only **5–10 samples**  
  - Others have **hundreds**  
- **Old datasets** sequenced on early R7.3 and R9 flow cells  
- Samples with **very low coverage** (< 3×)  
- Labels generated using **an outdated Kraken2 database**  
- No data cleaning of contaminants or host DNA

They train a neural network classifier using these data and deploy it for statewide pathogen surveillance.

## Case Study 2 – Vision AI for Disease Imaging (YOLO-Based Disease Detector)

A YOLO model is trained to detect **citrus canker** lesions on leaves.  
However, the training data have serious limitations:

- Only **one citrus cultivar** was used.  
- All images were taken on **sunny days** with uniform backgrounds.  
- Leaves were photographed at a **single angle** and distance.  
- Early lesions and mild symptoms were **poorly represented**.  

Despite this, the model is deployed across Florida and used by growers and extension agents in many different environments and cultivars.

## Case Study 3 – Sensitive Data Misuse in Agricultural AI (HLB Spread Modeling)

A team builds a model to predict **HLB (Huanglongbing) disease spread** in citrus.  
To improve performance, they collect and use **sensitive data**, including:

- **Exact GPS coordinates** of infected groves  
- **Proprietary orchard management data** from growers  
- Internal UF research datasets not meant for external sharing  
- ONT metagenomic metadata tied to **specific farms and growers**  

To make model development “easier,” a student uploads all these data to **public cloud AI tools** (e.g., general-purpose LLMs and external ML platforms) that are **not approved** for restricted or sensitive data.

No anonymization, aggregation, or de-identification was performed, and no data-classification check was done.

# Mapping the 5 Core Ethical Principles to the 3 Case Studies

This section shows how each case study relates to the **five ethical foundations** of responsible AI in plant pathology.

## Core Ethical Principles

1. **Transparency** – understanding how the model makes decisions  
2. **Fairness** – ensuring datasets represent all relevant groups/conditions  
3. **Accountability** – keeping humans responsible for decisions  
4. **Privacy** – protecting sensitive agricultural data  
5. **Safety** – validating and monitoring AI models for reliability

- **Case 2** violates **Fairness**, **Safety**, **Transparency**, and **Accountability** due to strong domain bias and lack of real-world validation.  
- **Case 3** critically violates **Privacy** and **Accountability**, and also harms **Safety** and **Transparency** because sensitive grower and location data are mishandled.
