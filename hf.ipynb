# Download dataSet from Hugging Face

```python
from datasets import load_dataset
ds = load_dataset("Project-AgML/apple_leaf_disease_classification")
```

The dataset card for "Project-AgML/apple_leaf_disease_classification" can be found on the Hugging Face Hub at: [https://huggingface.co/datasets/Project-AgML/apple_leaf_disease_classification](https://huggingface.co/datasets/Project-AgML/apple_leaf_disease_classification)

The whole data is loaded in ds

```python
ds
```

Visualize one element of the dataset

```python
import matplotlib.pyplot as plt
from PIL import Image

# Access the first example from the training split
example = ds['train'][0]

# Get the image and label
image = example['image']
label = example['label']

# Display the image
plt.imshow(image)
plt.title(f"Label: {label}")
plt.axis('off')
plt.show()
```

Your data in in memory no save yet.  Lest saved it

```python
import os

# Define the directory where you want to save the dataset
save_path = './apple_leaf_disease_dataset'

# Create the directory if it doesn't exist
os.makedirs(save_path, exist_ok=True)

# Save the dataset to disk
ds.save_to_disk(save_path)
print(f"Dataset saved to: {save_path}")
```

Compress

```python
import shutil

# Define the name for the zip file
zip_filename = 'apple_leaf_disease_dataset.zip'

# Create a zip archive of the saved dataset directory
shutil.make_archive(zip_filename.replace('.zip', ''), 'zip', save_path)

print(f"Dataset compressed to: {zip_filename}")
```

Lets extract the images. Remember the whole dataset is in different format

```python
import os

# Define the directory to save individual JPEG images
jpeg_output_dir = './apple_leaf_images_jpeg'
os.makedirs(jpeg_output_dir, exist_ok=True)

print(f"Saving images to: {jpeg_output_dir}")

# Iterate through the 'train' split and save each image
for i, example in enumerate(ds['train']):
    image = example['image']
    label = example['label'] # Assuming 'label' is an integer
    image_filename = os.path.join(jpeg_output_dir, f"image_{i}_label_{label}.jpg")
    image.save(image_filename, "JPEG")

print(f"Successfully saved {len(ds['train'])} images as JPEG files.")
```

Now that the individual JPEG images are saved in the `apple_leaf_images_jpeg` directory, you can compress this directory and download it to your local machine, similar to how we zipped the `save_path` directory earlier.

```python
import shutil

# Define the name for the zip file for JPEG images
jpeg_zip_filename = 'apple_leaf_images_jpeg.zip'

# Create a zip archive of the directory containing JPEG images
shutil.make_archive(jpeg_zip_filename.replace('.zip', ''), 'zip', jpeg_output_dir)

print(f"JPEG images compressed to: {jpeg_zip_filename}")
```

You will find `apple_leaf_images_jpeg.zip` in your Colab file explorer. Right-click on it and select 'Download' to get all the JPEG images on your local machine.

After running the cell above, you will see a file named `apple_leaf_disease_dataset.zip` in your Colab file explorer (look for the folder icon on the left sidebar). You can then right-click on this file and select 'Download' to transfer it to your local machine.
